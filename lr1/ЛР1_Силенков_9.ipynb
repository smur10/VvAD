```python
# переименование при необходимости
df_clean.columns = ['name', 'year', 'price', 'mileage', 'fuel', 'seller', 'transmission', 'owner']
```

```python
# Убираем выбросы в SellingPrice через IQR
Q1_price = df_clean['price'].quantile(0.25)
Q3_price = df_clean['price'].quantile(0.75)
IQR_price = Q3_price - Q1_price

df_clean = df_clean[(df_clean['price'] >= Q1_price - 1.5*IQR_price) &
                    (df_clean['price'] <= Q3_price + 1.5*IQR_price)]
```

```python
# Убираем выбросы в kmdriven через IQR
Q1_km = df_clean['mileage'].quantile(0.25)
Q3_km = df_clean['mileage'].quantile(0.75)
IQR_km = Q3_km - Q1_km

df_clean = df_clean[(df_clean['mileage'] >= Q1_km - 1.5*IQR_km) &
                    (df_clean['mileage'] <= Q3_km + 1.5*IQR_km)]
```

```python
print("Статистическое описание числовых признаков:")
display(df_clean.describe())
```

### 4. Проверка дубликатов

#### Проверка явных дубликатов

```python
print(df_clean.duplicated().sum())
```

```python
# удалите дубликаты, если они есть
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
```

#### Проверка неявных дубликатов

```python
for col in df_clean.columns:
    if df_clean[col].dtype == 'object':  # только текстовые столбцы
        # приводим строки к нижнему регистру и убираем пробелы
        cleaned = df_clean[col]
        
        # получаем уникальные значения
        unique_vals = cleaned.unique()
        
        print(f"\nСтолбец '{col}' - уникальные значения:")
        print(unique_vals)
```

```python
# удалите дубликаты, если они есть
df_clean['fuel'] = df_clean['fuel'].replace('Petroll', 'Petrol')
df_clean['transmission'] = df_clean['transmission'].replace('Manual', 'Mechanics')

duplicates = ['Dealeer', 'Trustmark Dealer'] # список неправильных названий
name = 'Dealer'# правильное название
df_clean['seller'] = df_clean['seller'].replace(duplicates, name)
```
